```python
#IMPORTING THE BASIC LIBRARY
import numpy as np
from matplotlib import pyplot as plt
```

```python
x=np.arange(1,11)
```

```python
y=2*x
```

LINE CHART

```python
plt.plot(x,y)
plt.title("line chart")
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.show()
```
